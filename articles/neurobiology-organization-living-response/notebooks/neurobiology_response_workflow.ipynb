This notebook scaffold supports the article **Neurobiology and the Organization of Living Response**. It can be expanded with leaky integration, threshold events, recurrent network response, neural-condition screening, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
params = pd.read_csv(article_dir / 'data' / 'neural_parameters.csv')
pulses = pd.read_csv(article_dir / 'data' / 'input_pulses.csv')
params, pulses
```

```python
param_dict = dict(zip(params['parameter'], params['value']))
dt = float(param_dict['dt'])
time = np.arange(0, float(param_dict['time_end']) + dt, dt)
inputs = np.zeros_like(time)

for _, pulse in pulses.iterrows():
    mask = (time >= pulse['start_time']) & (time < pulse['end_time'])
    inputs[mask] = pulse['input_amplitude']

voltage = np.zeros_like(time)
voltage[0] = float(param_dict['v_rest'])

for i in range(1, len(time)):
    d_voltage = (-(voltage[i - 1] - float(param_dict['v_rest'])) + inputs[i - 1]) / float(param_dict['tau'])
    voltage[i] = voltage[i - 1] + d_voltage * dt

output = pd.DataFrame({'time': time, 'input': inputs, 'voltage_state': voltage})
output.head().round(3)
```
