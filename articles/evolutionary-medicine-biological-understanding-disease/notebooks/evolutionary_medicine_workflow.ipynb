# Evolutionary Medicine and the Biological Understanding of Disease

Minimal notebook scaffold for evolutionary medicine workflows.

This notebook demonstrates conceptual modeling for evolutionary medicine. The data are synthetic and educational. The workflow is not suitable for clinical diagnosis, treatment, forecasting, or public-health decision-making.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
mismatch = pd.read_csv(article_dir / 'data' / 'mismatch_exposures.csv')
mismatch.head()
```

```python
mismatch['mismatch_distance'] = (mismatch['current_exposure'] - mismatch['adapted_exposure_reference']).abs()
mismatch['weighted_mismatch_score'] = mismatch['mismatch_distance'] * mismatch['evidence_confidence']
mismatch.sort_values('weighted_mismatch_score', ascending=False)
```

Interpretation warning: evolutionary explanations organize biological reasoning. They do not replace clinical evidence, public-health expertise, ethical judgment, or attention to social and environmental determinants of disease.
