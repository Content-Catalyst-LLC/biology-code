# Probability, Variation, and Biological Inference Workflow

This notebook scaffold supports binomial inference, Bayesian updating, bootstrap uncertainty, permutation testing, power simulation, likelihood comparison, false-discovery scaffolds, stochastic sampling, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
trials = pd.read_csv(article_dir / 'data' / 'binomial_trials.csv')
trials['estimate'] = trials['successes'] / trials['trials']
trials['standard_error'] = np.sqrt(trials['estimate'] * (1 - trials['estimate']) / trials['trials'])
trials.round(5)
```

```python
priors = pd.read_csv(article_dir / 'data' / 'bayesian_priors.csv')
priors['alpha_posterior'] = priors['alpha_prior'] + priors['successes']
priors['beta_posterior'] = priors['beta_prior'] + priors['trials'] - priors['successes']
priors['posterior_mean'] = priors['alpha_posterior'] / (priors['alpha_posterior'] + priors['beta_posterior'])
priors.round(5)
```

```python
measurements = pd.read_csv(article_dir / 'data' / 'biological_measurements.csv')
measurements.groupby('group').agg(n=('sample_id', 'count'), mean=('value', 'mean'), sd=('value', 'std')).round(5)
```

```python
pvals = pd.read_csv(article_dir / 'data' / 'multiple_testing.csv').sort_values('p_value')
n = len(pvals)
pvals['rank'] = np.arange(1, n + 1)
pvals['bh_raw'] = pvals['p_value'] * n / pvals['rank']
pvals['q_value_bh'] = np.minimum.accumulate(pvals['bh_raw'][::-1])[::-1]
pvals.round(5)
```
