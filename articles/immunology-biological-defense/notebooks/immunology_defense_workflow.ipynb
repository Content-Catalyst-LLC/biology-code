```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'immune_scenarios.csv')
thresholds = pd.read_csv(article_dir / 'data' / 'immune_thresholds.csv')
scenarios.head(), thresholds
```

```python
def simulate_scenario(P0=50, I0=2, D0=0, r=0.45, c=0.12, alpha=0.08, delta=0.18, gamma=0.06, rho=0.10, T=30, dt=0.05):
    time = np.arange(0, T + dt, dt)
    P = np.zeros(len(time))
    I = np.zeros(len(time))
    D = np.zeros(len(time))
    P[0], I[0], D[0] = P0, I0, D0
    for t in range(1, len(time)):
        dP = r * P[t - 1] - c * I[t - 1] * P[t - 1]
        dI = alpha * P[t - 1] - delta * I[t - 1]
        dD = gamma * I[t - 1] - rho * D[t - 1]
        P[t] = max(0, P[t - 1] + dP * dt)
        I[t] = max(0, I[t - 1] + dI * dt)
        D[t] = max(0, D[t - 1] + dD * dt)
    return pd.DataFrame({'time': time, 'pathogen_load': P, 'immune_activity': I, 'damage_burden': D})

series = simulate_scenario()
series.head().round(3)
```

```python
diagnostics = {
    'peak_pathogen': series['pathogen_load'].max(),
    'peak_immune': series['immune_activity'].max(),
    'peak_damage': series['damage_burden'].max(),
    'final_pathogen': series['pathogen_load'].iloc[-1],
    'final_damage': series['damage_burden'].iloc[-1]
}
pd.Series(diagnostics).round(3)
```
