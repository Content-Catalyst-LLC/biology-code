# Differential Equations in Population and Physiological Modeling Workflow

This notebook scaffold supports logistic growth, predator-prey dynamics, SIR epidemic dynamics, homeostatic regulation, pharmacokinetics, chemostat dynamics, reaction-diffusion scaffolds, sensitivity analysis, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
logistic = pd.read_csv(article_dir / 'data' / 'logistic_scenarios.csv')
logistic
```

```python
def logistic_final(N0, r, K, dt, t_end):
    time = np.arange(0, t_end + dt, dt)
    N = N0
    for _ in time[1:]:
        dN = r * N * (1 - N / K)
        N = max(N + dN * dt, 0)
    return N

rows = []
for _, s in logistic.iterrows():
    final = logistic_final(s.N0, s.r, s.K, s.dt, s.t_end)
    rows.append({'scenario': s.scenario, 'final_population': final, 'fraction_capacity': final / s.K})
pd.DataFrame(rows).round(5)
```
