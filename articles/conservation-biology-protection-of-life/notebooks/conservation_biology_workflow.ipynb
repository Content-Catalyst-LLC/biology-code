# Conservation Biology Workflow

This notebook scaffold supports the article **Conservation Biology and the Protection of Life**. It is intentionally minimal and can be expanded with population viability analysis, conservation priority scoring, sensitivity analysis, and provenance documentation.

```python
from pathlib import Path
import pandas as pd

article_dir = Path.cwd().parent
data_path = article_dir / 'data' / 'conservation_units.csv'
units = pd.read_csv(data_path)
units.head()
```

```python
weights = {
    'extinction_risk': 0.30,
    'endemism': 0.20,
    'habitat_loss': 0.20,
    'fragmentation': 0.15,
    'recovery_potential': 0.10,
    'cost_index': -0.05,
}

units['priority_score'] = sum(weights[col] * units[col] for col in weights)
units.sort_values('priority_score', ascending=False)
```
