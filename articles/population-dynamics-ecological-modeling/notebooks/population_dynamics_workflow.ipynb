```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'population_scenarios.csv')
scenarios.head()
```

```python
matrix_df = pd.read_csv(article_dir / 'data' / 'stage_matrix.csv')
stages = matrix_df['stage'].tolist()
A = matrix_df.drop(columns=['stage']).to_numpy(dtype=float)

initial_vector = pd.read_csv(article_dir / 'data' / 'initial_stage_vector.csv')['count'].to_numpy(dtype=float)

trajectory = [initial_vector]
for _ in range(20):
    trajectory.append(A @ trajectory[-1])

trajectory_df = pd.DataFrame(trajectory, columns=stages)
trajectory_df['total_population'] = trajectory_df.sum(axis=1)
trajectory_df.round(2)
```
