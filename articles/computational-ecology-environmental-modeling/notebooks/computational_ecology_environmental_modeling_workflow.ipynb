# Computational Ecology and Environmental Modeling Workflow

This notebook scaffold mirrors the repository workflow: habitat suitability, patch occupancy, environmental stress scenarios, runoff scaffolds, validation metrics, and provenance documentation.

```python
from pathlib import Path
import math
import pandas as pd

article_dir = Path.cwd().parent
sites = pd.read_csv(article_dir / 'data' / 'sites.csv')
sites.head()
```

```python
def logistic(x):
    return 1 / (1 + math.exp(-x))

def suitability(row):
    score = -2.0 + 0.05 * row['temperature_c'] + 0.0015 * row['precipitation_mm'] + 2.4 * row['habitat_quality'] - 2.0 * row['disturbance']
    return logistic(score)

sites['suitability'] = sites.apply(suitability, axis=1)
sites[['site_id', 'suitability', 'observed_presence']].round(5)
```

```python
scenarios = pd.read_csv(article_dir / 'data' / 'scenarios.csv')
scenarios['stress_index'] = 0.40 * scenarios['temperature_anomaly'] + 1.50 * scenarios['water_deficit'] + 1.20 * scenarios['disturbance'] - scenarios['habitat_gain']
scenarios['relative_resilience'] = 1 / (1 + scenarios['stress_index'])
scenarios[['scenario', 'stress_index', 'relative_resilience']].round(5)
```
