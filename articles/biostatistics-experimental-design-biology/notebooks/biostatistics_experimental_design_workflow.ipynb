# Biostatistics and Experimental Design in Biology Workflow

This notebook scaffold supports randomized allocation, blocked designs, two-group inference, power simulation, factorial design, nested replicate summaries, bootstrap uncertainty, permutation testing, assay-design summaries, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
data = pd.read_csv(article_dir / 'data' / 'two_group_measurements.csv')
data.groupby('group').agg(n=('value', 'count'), mean=('value', 'mean'), sd=('value', 'std')).round(5)
```

```python
control = data.loc[data['group'] == 'control', 'value'].to_numpy()
treated = data.loc[data['group'] == 'treated', 'value'].to_numpy()
n0, n1 = len(control), len(treated)
sd0, sd1 = control.std(ddof=1), treated.std(ddof=1)
pooled_sd = np.sqrt(((n0 - 1) * sd0**2 + (n1 - 1) * sd1**2) / (n0 + n1 - 2))
difference = treated.mean() - control.mean()
pd.DataFrame({'mean_difference': [difference], 'pooled_sd': [pooled_sd], 'effect_size_d': [difference / pooled_sd]}).round(5)
```

```python
blocked = pd.read_csv(article_dir / 'data' / 'blocked_design.csv')
wide = blocked.pivot(index='block', columns='treatment', values='response')
wide['within_block_difference'] = wide['treated'] - wide['control']
wide.round(5)
```

```python
factorial = pd.read_csv(article_dir / 'data' / 'factorial_design_observations.csv')
factorial.groupby(['temperature', 'nutrient']).agg(n=('response', 'count'), mean_response=('response', 'mean'), sd_response=('response', 'std')).round(5)
```
