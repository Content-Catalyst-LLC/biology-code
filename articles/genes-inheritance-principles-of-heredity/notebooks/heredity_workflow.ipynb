# Genes, Inheritance, and the Principles of Heredity Workflow

This notebook scaffold supports the article **Genes, Inheritance, and the Principles of Heredity**. It can be expanded with genotype expectations, allele-frequency estimation, chi-square testing, recombination, quantitative-trait scaffolds, heredity condition scoring, and provenance notes.

```python
from pathlib import Path
import pandas as pd
import numpy as np

article_dir = Path.cwd().parent
geno = pd.read_csv(article_dir / 'data' / 'genotype_counts.csv')
counts = dict(zip(geno['genotype'], geno['count']))
n_AA = counts.get('AA', 0)
n_Aa = counts.get('Aa', 0)
n_aa = counts.get('aa', 0)
n = n_AA + n_Aa + n_aa
p = (2 * n_AA + n_Aa) / (2 * n)
q = 1 - p
pd.DataFrame({'n':[n], 'p':[p], 'q':[q], 'expected_heterozygosity':[2*p*q]}).round(4)
```

```python
mono = pd.read_csv(article_dir / 'data' / 'monohybrid_counts.csv')
total = mono['observed'].sum()
mono['expected'] = total * mono['expected_ratio']
mono['chi_component'] = (mono['observed'] - mono['expected'])**2 / mono['expected']
mono.round(4)
```

```python
recomb = pd.read_csv(article_dir / 'data' / 'recombination_observations.csv')
estimated_r = recomb.loc[recomb['class'] == 'recombinant', 'count'].sum() / recomb['count'].sum()
pd.DataFrame({'estimated_recombination_fraction':[estimated_r]}).round(4)
```

```python
trait = pd.read_csv(article_dir / 'data' / 'quantitative_trait.csv')
VA = trait['additive_genetic_value'].var(ddof=1)
VP = trait['phenotype'].var(ddof=1)
h2 = VA / VP
pd.DataFrame({'additive_variance':[VA], 'phenotypic_variance':[VP], 'h2':[h2]}).round(4)
```

```python
condition = pd.read_csv(article_dir / 'data' / 'heredity_condition_sites.csv')
condition['heredity_condition_score'] = (
    0.18 * condition['standing_variation'] +
    0.14 * condition['inheritance_clarity'] +
    0.12 * condition['recombination_information'] +
    0.15 * condition['population_size'] +
    0.15 * condition['genotype_quality'] +
    0.14 * condition['environmental_context'] +
    0.12 * (1 - condition['inbreeding_risk'])
)
condition.sort_values('heredity_condition_score', ascending=False).round(3)
```
