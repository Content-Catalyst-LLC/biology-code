# Biosphere and Planetary Life Support Systems Workflow

This notebook scaffold supports the article **The Biosphere and Planetary Life Support Systems**. It is intentionally minimal and can be expanded with functional-integrity scoring, carbon-balance scenarios, disturbance sensitivity analysis, and provenance documentation.

```python
from pathlib import Path
import pandas as pd

article_dir = Path.cwd().parent
data_path = article_dir / 'data' / 'biosphere_units.csv'
units = pd.read_csv(data_path)
units.head()
```

```python
units['functional_integrity'] = (
    0.20 * units['primary_production'] +
    0.18 * units['water_regulation'] +
    0.18 * units['nutrient_retention'] +
    0.18 * units['habitat_complexity'] +
    0.12 * units['connectivity'] +
    0.14 * units['biodiversity_signal'] -
    0.20 * units['disturbance_pressure']
)

units.sort_values('functional_integrity', ascending=False)
```
