# Speciation, Diversity, and the Tree of Life Workflow

This notebook scaffold supports the article **Speciation, Diversity, and the Tree of Life**. It can be expanded with divergence-with-gene-flow simulation, FST-style structure, sequence-distance matrices, birth-death diversification, speciation condition scoring, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'divergence_scenarios.csv')
scenarios
```

```python
seqs = pd.read_csv(article_dir / 'data' / 'sequences.csv')
seqs['sequence_length'] = seqs['sequence'].str.len()
seqs
```

```python
bd = pd.read_csv(article_dir / 'data' / 'birth_death_scenarios.csv')
bd['net_diversification'] = bd['lambda_rate'] - bd['mu_rate']
bd['turnover'] = bd['lambda_rate'] + bd['mu_rate']
bd.sort_values('net_diversification', ascending=False).round(4)
```

```python
condition = pd.read_csv(article_dir / 'data' / 'speciation_condition_sites.csv')
condition['speciation_condition_score'] = (
    0.20 * condition['allele_divergence'] +
    0.20 * condition['reproductive_isolation'] +
    0.18 * condition['ecological_difference'] +
    0.16 * condition['phylogenetic_resolution'] +
    0.14 * (1 - condition['gene_flow_risk']) +
    0.12 * condition['lineage_distinctiveness']
)
condition.sort_values('speciation_condition_score', ascending=False).round(3)
```
