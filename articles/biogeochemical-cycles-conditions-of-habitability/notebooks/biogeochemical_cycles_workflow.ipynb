# Biogeochemical Cycles and Habitability Workflow

This notebook scaffold supports the article **Biogeochemical Cycles and the Conditions of Habitability**. It is intentionally minimal and can be expanded with mass-balance modeling, habitability-support scoring, oxygen-stress screening, nutrient-loading scenarios, and provenance documentation.

```python
from pathlib import Path
import pandas as pd

article_dir = Path.cwd().parent
units = pd.read_csv(article_dir / 'data' / 'habitability_units.csv')
units.head()
```

```python
units['habitability_support'] = (
    0.18 * units['carbon_uptake_capacity'] +
    0.16 * units['water_regulation'] +
    0.14 * units['nitrogen_retention'] +
    0.14 * units['phosphorus_buffering'] +
    0.16 * units['oxygen_stability'] -
    0.10 * units['disturbance_pressure'] -
    0.06 * units['acidification_pressure'] -
    0.06 * units['nutrient_loading']
)

units.sort_values('habitability_support', ascending=False)
```
