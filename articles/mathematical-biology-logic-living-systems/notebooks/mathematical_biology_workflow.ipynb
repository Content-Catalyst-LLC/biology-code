# Mathematical Biology and the Logic of Living Systems Workflow

This notebook scaffold supports population dynamics, predator-prey interaction, SIR epidemic dynamics, enzyme kinetics, stochastic simulation, network analysis, sensitivity, optimization, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
logistic = pd.read_csv(article_dir / 'data' / 'logistic_scenarios.csv')
logistic
```

```python
def logistic_growth(time, N0, r, K):
    return K / (1 + ((K - N0) / N0) * np.exp(-r * time))

rows = []
for _, s in logistic.iterrows():
    time = np.arange(0, s['time_end'] + s['dt'], s['dt'])
    population = logistic_growth(time, s['initial_population'], s['growth_rate'], s['carrying_capacity'])
    rows.append({'scenario': s['scenario'], 'final_population': population[-1], 'fraction_capacity': population[-1] / s['carrying_capacity']})

pd.DataFrame(rows).round(5)
```

```python
sir = pd.read_csv(article_dir / 'data' / 'sir_scenarios.csv')
sir['R0'] = sir['beta'] / sir['gamma']
sir[['scenario', 'beta', 'gamma', 'R0']].round(4)
```

```python
enzyme = pd.read_csv(article_dir / 'data' / 'enzyme_kinetics.csv')
enzyme['velocity'] = enzyme['Vmax'] * enzyme['substrate'] / (enzyme['Km'] + enzyme['substrate'])
enzyme.round(4)
```

```python
edges = pd.read_csv(article_dir / 'data' / 'network_edges.csv')
nodes = sorted(set(edges['source']).union(edges['target']))
degree = []
for node in nodes:
    degree.append({'node': node, 'in_degree': int((edges['target'] == node).sum()), 'out_degree': int((edges['source'] == node).sum())})
pd.DataFrame(degree)
```
