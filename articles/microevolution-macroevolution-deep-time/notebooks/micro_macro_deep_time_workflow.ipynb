# Microevolution, Macroevolution, and Deep Time Workflow

This notebook scaffold supports the article **Microevolution, Macroevolution, and Deep Time**. It can be expanded with Hardy-Weinberg expectations, Wright-Fisher drift, sequence divergence, clade turnover, birth-death simulation, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'population_scenarios.csv')
scenarios['q0'] = 1 - scenarios['p0']
scenarios['expected_AA'] = scenarios['p0'] ** 2
scenarios['expected_Aa'] = 2 * scenarios['p0'] * scenarios['q0']
scenarios['expected_aa'] = scenarios['q0'] ** 2
scenarios[['scenario', 'p0', 'q0', 'expected_AA', 'expected_Aa', 'expected_aa']].round(4)
```

```python
clades = pd.read_csv(article_dir / 'data' / 'clade_turnover.csv')
clades['lambda_rate'] = clades['originations'] / clades['interval_myr']
clades['mu_rate'] = clades['extinctions'] / clades['interval_myr']
clades['net_diversification'] = clades['lambda_rate'] - clades['mu_rate']
clades['turnover'] = clades['lambda_rate'] + clades['mu_rate']
clades.round(4)
```

```python
scale = pd.read_csv(article_dir / 'data' / 'evolutionary_scale_sites.csv')
scale['evolutionary_scale_score'] = (
    0.18 * scale['population_variation'] +
    0.18 * scale['lineage_distinctiveness'] +
    0.16 * scale['fossil_record_strength'] +
    0.16 * scale['phylogenetic_resolution'] +
    0.16 * (1 - scale['extinction_pressure']) +
    0.16 * scale['adaptive_capacity']
)
scale.sort_values('evolutionary_scale_score', ascending=False).round(3)
```
