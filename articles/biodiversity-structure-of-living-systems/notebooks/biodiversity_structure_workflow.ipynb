# Biodiversity and the Structure of Living Systems Workflow

This notebook scaffold supports the article **Biodiversity and the Structure of Living Systems**. It is intentionally minimal and can be expanded with diversity metrics, beta diversity, ordination, trait-based analysis, priority screening, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
community = pd.read_csv(article_dir / 'data' / 'community_matrix.csv').set_index('site')
community.head()
```

```python
relative_abundance = community.div(community.sum(axis=1), axis=0)
safe_relative_abundance = relative_abundance.replace(0, np.nan)

shannon = -(
    safe_relative_abundance * np.log(safe_relative_abundance)
).sum(axis=1).fillna(0)

summary = pd.DataFrame({
    'richness': (community > 0).sum(axis=1),
    'shannon': shannon,
    'hill_q1': np.exp(shannon),
    'hill_q2': 1 / (relative_abundance ** 2).sum(axis=1)
})

summary.round(3)
```
