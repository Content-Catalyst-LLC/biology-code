```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
traits = pd.read_csv(article_dir / 'data' / 'species_traits.csv')
traits.head()
```

```python
B0 = 4.2
traits['metabolic_rate'] = B0 * traits['body_mass_kg'] ** 0.75
traits['mass_specific_rate'] = traits['metabolic_rate'] / traits['body_mass_kg']
traits[['species', 'body_mass_kg', 'metabolic_rate', 'mass_specific_rate']].round(4)
```

```python
A = np.array([[0.0, 1.4], [0.35, 0.72]])
n = np.array([40.0, 25.0])
trajectory = [n.copy()]

for _ in range(20):
    n = A @ n
    trajectory.append(n.copy())

projection = pd.DataFrame(trajectory, columns=['juveniles', 'adults'])
projection['year'] = range(21)
projection['total_population'] = projection['juveniles'] + projection['adults']
projection.round(2).head()
```
