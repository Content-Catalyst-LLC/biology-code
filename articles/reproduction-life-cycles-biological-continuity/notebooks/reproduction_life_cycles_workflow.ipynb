```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
matrix_df = pd.read_csv(article_dir / 'data' / 'stage_matrix.csv')
stages = matrix_df['stage'].tolist()
A = matrix_df.drop(columns=['stage']).to_numpy(dtype=float)

initial = pd.read_csv(article_dir / 'data' / 'initial_stage_vector.csv')['count'].to_numpy(dtype=float)
A, initial
```

```python
trajectory = [initial]
for _ in range(20):
    trajectory.append(A @ trajectory[-1])

trajectory_df = pd.DataFrame(trajectory, columns=stages)
trajectory_df['total'] = trajectory_df.sum(axis=1)
trajectory_df.round(2)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
dominant_index = np.argmax(np.real(eigenvalues))
lambda_dominant = np.real(eigenvalues[dominant_index])

stable_stage = np.real(eigenvectors[:, dominant_index])
stable_stage = stable_stage / stable_stage.sum()

pd.DataFrame({
    'stage': stages,
    'stable_stage_distribution': stable_stage
}).round(4), lambda_dominant
```
