# Evolution and the History of Life Workflow

This notebook scaffold supports the article **Evolution and the History of Life**. It can be expanded with allele-frequency evolution, sequence divergence, birth-death diversification, major-transition documentation, evolutionary condition scoring, and provenance notes.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'evolutionary_scenarios.csv')
p = scenarios['p0']
q = 1 - p
scenarios['expected_AA'] = p ** 2
scenarios['expected_Aa'] = 2 * p * q
scenarios['expected_aa'] = q ** 2
scenarios[['scenario', 'p0', 'expected_AA', 'expected_Aa', 'expected_aa']].round(4)
```

```python
seqs = pd.read_csv(article_dir / 'data' / 'sequences.csv')
seqs['sequence_length'] = seqs['sequence'].str.len()
seqs
```

```python
bd = pd.read_csv(article_dir / 'data' / 'birth_death_scenarios.csv')
bd['net_diversification'] = bd['lambda_rate'] - bd['mu_rate']
bd['turnover'] = bd['lambda_rate'] + bd['mu_rate']
bd.sort_values('net_diversification', ascending=False).round(4)
```

```python
transitions = pd.read_csv(article_dir / 'data' / 'major_transitions.csv')
transitions
```

```python
condition = pd.read_csv(article_dir / 'data' / 'evolutionary_condition_sites.csv')
condition['evolutionary_condition_score'] = (
    0.17 * condition['standing_variation'] +
    0.17 * condition['phylogenetic_signal'] +
    0.16 * condition['fossil_record_strength'] +
    0.16 * (1 - condition['environmental_change']) +
    0.17 * (1 - condition['extinction_pressure']) +
    0.17 * condition['adaptive_capacity']
)
condition.sort_values('evolutionary_condition_score', ascending=False).round(3)
```
