# Synthetic Biology and the Engineering of Biological Systems

Minimal notebook scaffold for synthetic biology design-build-test-learn workflows.

This notebook demonstrates conceptual scoring for engineered biological systems. The data are synthetic and educational. The workflow is not suitable for organism construction, biosafety approval, regulatory submission, or environmental release decisions.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
designs = pd.read_csv(article_dir / 'data' / 'synthetic_biology_designs.csv')
designs.head()
```

```python
designs['engineering_score'] = (
    designs['output_signal'] * 0.40
    + designs['genetic_stability'] * 0.30
    - designs['host_burden'] * 0.20
    - designs['measurement_uncertainty'] * 0.10
)
designs.sort_values('engineering_score', ascending=False)
```

Interpretation warning: numerical scores organize assumptions. They do not replace sequence verification, calibrated measurement, biological replication, biosafety review, biosecurity review, or responsible governance.
