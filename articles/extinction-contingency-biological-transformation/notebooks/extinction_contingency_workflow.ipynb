# Extinction, Contingency, and Biological Transformation Workflow

This notebook scaffold supports the article **Extinction, Contingency, and Biological Transformation**. It can be expanded with survivorship proportions, hazard models, stochastic survival, post-crisis recovery, trait-risk screening, phylogenetic-loss scoring, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
clades = pd.read_csv(article_dir / 'data' / 'clade_survivorship.csv')
clades['survivorship'] = clades['survivors'] / clades['initial']
clades['extinction'] = 1 - clades['survivorship']
clades.round(4)
```

```python
hazards = pd.read_csv(article_dir / 'data' / 'hazard_scenarios.csv')
hazards['survivorship_at_horizon'] = np.exp(-hazards['lambda'] * hazards['time_horizon'])
hazards.round(4)
```

```python
traits = pd.read_csv(article_dir / 'data' / 'trait_risk_taxa.csv')
traits['risk_index'] = (
    0.4 * (1 - traits['range_size']) +
    0.3 * (1 - traits['trophic_flexibility']) +
    0.3 * traits['habitat_dependence']
)
traits.sort_values('risk_index', ascending=False).round(3)
```

```python
phylo = pd.read_csv(article_dir / 'data' / 'phylogenetic_loss.csv')
total_history = phylo['branch_length'].sum()
lost_history = phylo.loc[phylo['status'] == 'extinct', 'branch_length'].sum()
lost_history / total_history
```
