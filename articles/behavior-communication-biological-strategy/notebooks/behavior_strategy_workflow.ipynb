```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
options = pd.read_csv(article_dir / 'data' / 'behavioral_options.csv')
options
```

```python
def softmax(values, beta=1.1):
    centered = values - values.max()
    ex = np.exp(beta * centered)
    return ex / ex.sum()

options['utility'] = (
    options['benefit']
    - 0.8 * options['energetic_cost']
    - 1.2 * options['predation_risk']
)

options['choice_probability'] = softmax(options['utility'])
options.round(3)
```

```python
signals = pd.read_csv(article_dir / 'data' / 'signaling_strategies.csv')
receiver_state = 0.75

signals['sender_utility'] = (
    signals['mate_benefit']
    - 0.8 * signals['energetic_cost']
    - 1.1 * signals['predator_exposure']
)

signals['receiver_response'] = 1 / (
    1 + np.exp(-6 * (signals['receiver_detectability'] * receiver_state - 0.35))
)

signals['combined_score'] = signals['sender_utility'] * signals['receiver_response']
signals.round(3)
```
