# Agriculture, Food Systems, and the Management of Life

Minimal notebook scaffold for agriculture and food-system indicators.

This notebook demonstrates conceptual food-system indicators using synthetic data. The workflow is educational and not suitable for farm-management, nutrition, policy, or climate-risk decisions.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
systems = pd.read_csv(article_dir / 'data' / 'production_systems.csv')
systems.head()
```

```python
systems['yield_t_ha'] = systems['production_tonnes'] / systems['area_hectares']
systems['water_productivity_t_per_m3'] = systems['production_tonnes'] / systems['water_used_m3']
systems['nutrient_use_efficiency'] = systems['nutrient_harvested_kg'] / systems['nutrient_input_kg']
systems['food_loss_rate'] = systems['food_lost_tonnes'] / systems['production_tonnes']
systems.sort_values('yield_t_ha', ascending=False)
```

Interpretation warning: indicators organize assumptions. They do not replace field data, farmer knowledge, ecological expertise, nutrition assessment, public-health review, or governance judgment.
