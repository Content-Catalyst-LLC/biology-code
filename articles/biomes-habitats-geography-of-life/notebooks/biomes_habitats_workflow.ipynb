# Biomes, Habitats, and the Geography of Life Workflow

This notebook scaffold supports the article **Biomes, Habitats, and the Geography of Life**. It is intentionally minimal and can be expanded with species-area modeling, habitat suitability, spatial prioritization, and provenance documentation.

```python
from pathlib import Path
import pandas as pd

article_dir = Path.cwd().parent
sites = pd.read_csv(article_dir / 'data' / 'habitat_sites.csv')
sites.head()
```

```python
sites['spatial_priority'] = (
    0.25 * (sites['precipitation'] / sites['precipitation'].max()) +
    0.25 * sites['soil_quality'] +
    0.25 * sites['connectivity'] -
    0.15 * sites['disturbance'] -
    0.10 * sites['land_use_pressure']
)

sites.sort_values('spatial_priority', ascending=False)
```
