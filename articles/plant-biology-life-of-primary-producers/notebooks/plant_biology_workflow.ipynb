# Plant Biology and the Life of Primary Producers Workflow

This notebook scaffold supports the article **Plant Biology and the Life of Primary Producers**. It can be expanded with productivity accounting, light-response curves, biomass recovery, plant condition scoring, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
sites = pd.read_csv(article_dir / 'data' / 'productivity_sites.csv')
sites['NPP'] = sites['GPP'] - sites['Ra']
sites['NEP'] = sites['GPP'] - (sites['Ra'] + sites['Rh'])
sites
```

```python
def light_response(I, alpha=0.05, Amax=18, Rd=1.5):
    return (alpha * I * Amax) / (alpha * I + Amax) - Rd

scenarios = pd.read_csv(article_dir / 'data' / 'light_response_scenarios.csv')
irradiance = np.arange(0, 2001, 100)
rows = []

for _, row in scenarios.iterrows():
    for I in irradiance:
        rows.append({
            'scenario': row['scenario'],
            'irradiance': I,
            'assimilation': light_response(I, row['alpha'], row['Amax'], row['Rd'])
        })

light_df = pd.DataFrame(rows)
light_df.head()
```

```python
condition = pd.read_csv(article_dir / 'data' / 'plant_condition_sites.csv')
condition['plant_condition_score'] = (
    0.20 * condition['canopy_condition'] +
    0.18 * condition['water_availability'] +
    0.16 * condition['nutrient_status'] +
    0.16 * condition['soil_function'] +
    0.15 * condition['regeneration_support'] +
    0.08 * (1 - condition['disease_pressure']) +
    0.07 * (1 - condition['drought_stress'])
)
condition.sort_values('plant_condition_score', ascending=False).round(3)
```
