# R for Biological Data Analysis and Visualization Workflow

This notebook scaffold mirrors the R-first workflow: biological measurement summaries, ecological diversity, dose-response exploration, metadata validation, and reproducibility documentation.

```python
from pathlib import Path
import pandas as pd
import numpy as np

article_dir = Path.cwd().parent
measurements = pd.read_csv(article_dir / 'data' / 'measurements.csv')
measurements.head()
```

```python
valid = measurements[(measurements['qc_flag'] == 'pass') & measurements['value'].notna()]
summary = valid.groupby('treatment')['value'].agg(['count', 'mean', 'std', 'min', 'max'])
summary['se'] = summary['std'] / np.sqrt(summary['count'])
summary['cv'] = summary['std'] / summary['mean']
summary.round(5)
```

```python
species = pd.read_csv(article_dir / 'data' / 'species_counts.csv')
def shannon(counts):
    counts = np.array([c for c in counts if c > 0], dtype=float)
    p = counts / counts.sum()
    return float(-(p * np.log(p)).sum())

eco = species.groupby(['site', 'habitat']).agg(total_abundance=('count', 'sum'), richness=('count', lambda x: (x > 0).sum()), shannon=('count', shannon)).reset_index()
eco.round(5)
```
