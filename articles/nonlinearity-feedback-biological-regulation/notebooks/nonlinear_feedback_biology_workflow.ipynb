# Nonlinearity, Feedback, and Biological Regulation Workflow

This notebook scaffold supports saturating response, Hill functions, negative feedback, positive feedback, delayed feedback, logistic regulation, sensitivity analysis, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
sat = pd.read_csv(article_dir / 'data' / 'saturating_response_scenarios.csv')
sat['response_at_20'] = sat['vmax'] * 20 / (sat['k_half'] + 20)
sat['response_at_80'] = sat['vmax'] * 80 / (sat['k_half'] + 80)
sat.round(5)
```

```python
hill = pd.read_csv(article_dir / 'data' / 'hill_scenarios.csv')
signal = 60
hill['response_at_60'] = signal ** hill['hill_coefficient'] / (hill['k_half'] ** hill['hill_coefficient'] + signal ** hill['hill_coefficient'])
hill.round(5)
```
