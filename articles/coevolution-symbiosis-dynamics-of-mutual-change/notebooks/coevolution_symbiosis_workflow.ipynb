This notebook scaffold supports the article **Coevolution, Symbiosis, and the Dynamics of Mutual Change**. It can be expanded with benefit-cost thresholds, reciprocal feedback, host-pathogen arms-race simulation, interaction-network dependency scoring, and provenance documentation.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path.cwd().parent
scenarios = pd.read_csv(article_dir / 'data' / 'benefit_cost_scenarios.csv')
scenarios['benefit'] = scenarios['benefit_intercept'] - scenarios['benefit_stress_slope'] * scenarios['stress']
scenarios['cost'] = scenarios['cost_intercept'] + scenarios['cost_stress_slope'] * scenarios['stress']
scenarios['net_effect'] = scenarios['symbiont_load'] * (scenarios['benefit'] - scenarios['cost'])
scenarios[['scenario', 'stress', 'benefit', 'cost', 'net_effect']].round(3)
```

```python
network = pd.read_csv(article_dir / 'data' / 'network_interactions.csv')
network['weighted_support'] = network['weight'] * network['partner_reliability']
dependency = network.groupby('focal').agg(
    dependency_support=('weighted_support', 'sum'),
    partner_count=('partner', 'count')
).reset_index()
dependency.round(3)
```

```python
condition = pd.read_csv(article_dir / 'data' / 'symbiosis_condition_sites.csv')
condition['symbiosis_condition_score'] = (
    0.22 * condition['partner_presence'] +
    0.22 * condition['interaction_stability'] +
    0.18 * condition['transmission_reliability'] +
    0.16 * condition['functional_redundancy'] +
    0.12 * (1 - condition['environmental_stress']) +
    0.10 * (1 - condition['cheating_pressure'])
)
condition.sort_values('symbiosis_condition_score', ascending=False).round(3)
```
